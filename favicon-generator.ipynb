```python
# Notebook para gerar um favicon.ico a partir da imagem logo.jpeg e utilizá-lo em uma aplicação web.

## 1. Importar bibliotecas
from pathlib import Path
from PIL import Image

## 2. Carregar a imagem logo.jpeg
root = Path.cwd()
source = root / 'public' / 'images' / 'logo.jpeg'
print(f'Arquivo de origem: {source}')
print(f'Existe: {source.exists()}')

if not source.exists():
    raise FileNotFoundError(f'Arquivo não encontrado: {source}')

img = Image.open(source)
print(f'Formato: {img.format}')
print(f'Tamanho: {img.size}')
print(f'Modo: {img.mode}')

## 3. Validar e analisar as dimensões da imagem
width, height = img.size
if width <= 0 or height <= 0:
    raise ValueError('Imagem inválida: dimensões fora do intervalo esperado.')

print(f'Imagem válida. Dimensões: {width}x{height}')

## 4. Redimensionar e preparar o ícone
sizes = [16, 32, 48, 64, 128, 256]
icons = []
for size in sizes:
    thumb = img.resize((size, size), Image.Resampling.LANCZOS).convert('RGBA')
    icons.append(thumb)

print('Ícones preparados para os tamanhos:', sizes)

## 5. Gerar o arquivo favicon.ico
output = root / 'public' / 'favicon.ico'
# Pillow salva o arquivo ICO a partir de uma imagem RGBA
icons[0].save(output, format='ICO', sizes=[(s, s) for s in sizes])
print(f'Arquivo salvo em: {output}')

## 6. Salvar em diretórios do projeto
# O favicon foi salvo em public/favicon.ico para uso em aplicações web.
print('Diretório correto: public/favicon.ico')

## 7. Testar o favicon na aplicação
print('Favicon pronto para ser usado em uma aplicação web, por exemplo em <head><link rel="icon" href="/favicon.ico"></head>.')
```
